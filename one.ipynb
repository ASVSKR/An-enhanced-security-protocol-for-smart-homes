```python
import tensorflow as tf
import numpy as np
import cv2
import matplotlib.pyplot as plt
import FingerBiometrics as FB

# Step 1: Function to convert image to bytes
def image_to_bytes(image_path):
    # Read the image from the file
    img = cv2.imread(image_path)

    # Encode the image into a byte array (memory buffer)
    _, img_bytes = cv2.imencode('.jpg', img)  # Change the extension to '.png' if needed

    # Convert the buffer to bytes
    img_bytes = img_bytes.tobytes()

    return img_bytes

# Function to preprocess the image (same as before)
def preprocess_image_refined(img):
    # Convert to grayscale
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # Histogram equalization
    img_eq = cv2.equalizeHist(img_gray)
    
    # Gaussian Blur for noise reduction
    img_blurred = cv2.GaussianBlur(img_eq, (5, 5), 0)
    
    # Optional: Resize to standard size
    img_resized = cv2.resize(img_blurred, (600, 600))
    
    return img_resized

def fingerDetection(image_path, model):
    img_bytes = image_to_bytes(image_path)
    # Now load the image from bytes
    image_np = FB.load_image_into_numpy_array(img_bytes)

    detections = FB.detect_objects(image_np, model)

    boxes = detections['detection_boxes'][0].numpy()
    scores = detections['detection_scores'][0].numpy()
    classes = detections['detection_classes'][0].numpy().astype(np.int32)

    cropped_image, image_with_boxes = FB.draw_best_box(image_np, boxes, scores, classes)

    print("Detection and cropping completed.")
    cropped_image = preprocess_image_refined(cropped_image)

    return cropped_image
```

```python
# Function to perform AKAZE feature matching
def feature_matching_AKAZE(img1, img2, threshold=0.3):
    # Create AKAZE detector
    akaze = cv2.AKAZE_create()

    # Detect keypoints and compute descriptors
    kp1, des1 = akaze.detectAndCompute(img1, None)
    kp2, des2 = akaze.detectAndCompute(img2, None)

    # Brute Force Matcher
    bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)
    matches = bf.match(des1, des2)

    # Sort the matches based on distance
    matches = sorted(matches, key=lambda x: x.distance)

    print(f"Total Matches = {len(kp1)}, {len(kp2)}")
    print(len(matches))

    # Filter matches using a threshold
    good_matches = [m for m in matches if m.distance < threshold]
    print(f"AKAZE matches: {len(good_matches)}")

    return len(good_matches)

# Function to perform ORB feature matching
def feature_matching_ORB(img1, img2, threshold=0.3):
    # Create ORB detector
    orb = cv2.ORB_create()

    # Detect keypoints and compute descriptors
    kp1, des1 = orb.detectAndCompute(img1, None)
    kp2, des2 = orb.detectAndCompute(img2, None)

    # Brute Force Matcher with Hamming distance (for binary descriptors)
    bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
    matches = bf.match(des1, des2)

    # Sort the matches based on distance
    matches = sorted(matches, key=lambda x: x.distance)

    print(f"Total Matches = {len(kp1)}, {len(kp2)}")
    print(len(matches))

    # Filter matches using a threshold
    good_matches = [m for m in matches if m.distance < threshold]
    print(f"ORB matches: {len(good_matches)}")

    return len(good_matches)

# Function to perform SIFT feature matching
def feature_matching_SIFT(img1, img2, threshold=0.3):
    # Create SIFT detector
    sift = cv2.SIFT_create()

    # Detect keypoints and compute descriptors
    kp1, des1 = sift.detectAndCompute(img1, None)
    kp2, des2 = sift.detectAndCompute(img2, None)

    # Brute Force Matcher
    bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)
    matches = bf.match(des1, des2)

    # Sort the matches based on distance (smallest distance first)
    matches = sorted(matches, key=lambda x: x.distance)
    print(f"Total Matches = {len(kp1)}, {len(kp2)}")
    print(len(matches))
    # Filter matches using a threshold
    good_matches = [m for m in matches if m.distance < threshold]
    print(f"SIFT matches: {len(good_matches)}")

    return len(good_matches)
```

```python
# isolate the finger from the image

image_path = ["5.jpg", "6.jpg"]
model = tf.saved_model.load('./Model/Fingerprint/ssd_mobilenet_v2_320x320_coco17_tpu-8/saved_model')
savedimages = []
print(len(image_path))
for i in range(len(image_path)):
    print(f"i = {i}")
    savedimages.append(fingerDetection(image_path[i], model))

print(len(savedimages))
for i in range(len(savedimages)):
    if savedimages[i] is not None:
        plt.figure(figsize=(6, 6))
        plt.imshow(savedimages[i])
        plt.axis("off")  # Hide axes
        plt.title("Best Detected Fingerprint")
        plt.show()
    else:
        print("❌ No valid cropped image found.")
        
matching = []
matching.append(feature_matching_AKAZE(savedimages[0], savedimages[1], threshold=0.3))
matching.append(feature_matching_ORB(savedimages[0], savedimages[1], threshold=0.3))
matching.append(feature_matching_SIFT(savedimages[0], savedimages[1], threshold=0.3))
```
